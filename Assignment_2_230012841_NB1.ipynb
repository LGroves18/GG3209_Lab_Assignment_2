### 1.8 Create a script that generates a multiplication table:

* Write a function that takes an integer as input and generates a multiplication table for that number, from 1 to 10.
* Call the function with a few different numbers and print the results.
* Optionally, you can use Markdown to format the output as a table.

````python
def multiplication_table(num):
    print("--- Table for", num, "---")
    for i in range(1, 11):
        print(num, "x", i, "=", num * i)
    print()

multiplication_table(5)
multiplication_table(7)
````

## 2.15

What is the total emissions of all the meat products (Pork, Poultry, Fish, Lamb & Goat, Beef) in the dataset combined?

````python
# 1st step - define the categories
meat_products = ['Pork', 'Poultry', 'Fish', 'Lamb & Goat', 'Beef']
print(meat_products)
````

````python
#2nd step - filter the dataframe. 
meat_df = df[df['food_category'].isin(meat_products)]
print(meat_df.head(5))
````

````python
#3rd step - sum the emissions
total_meat_co2 = meat_df['co2_emmission'].sum()
print("Total CO2 emissions for all meat products combined:")
print(total_meat_co2)
````

# 2.16 

What is the total emissions of all other (non-meat) products in the dataset combined?

````python
meat_products = ['Pork', 'Poultry', 'Fish', 'Lamb & Goat', 'Beef']
print(meat_products)

non_meat_df = df[~df['food_category'].isin(meat_products)]
print(non_meat_df.head(5))

#3rd step - sum the emissions
total_non_meat_co2 = non_meat_df['co2_emmission'].sum()
print("Total CO2 emissions for all meat products combined:")
print(total_non_meat_co2)

the_total = df["co2_emmission"].sum()
print(the_total)

#JUST CHECKING AND IT's CORRECT!!!! YAY
print(the_total - total_non_meat_co2)
````
